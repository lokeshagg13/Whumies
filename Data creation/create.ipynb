```python
import random
import string

# Suppose you already have your list of 2000 first names
with open("Indian Names/first_name.txt", "r") as f:
    first_names = f.read()   # reads entire file into a string
first_names = first_names.split('\n')

# Convert names to title case for consistency
first_names = [name.title() for name in first_names]

# Step 1: Ensure at least one name per alphabet
selected = []
for letter in string.ascii_uppercase:   # 'A' through 'Z'
    # Find all names starting with that letter
    candidates = [name for name in first_names if name.startswith(letter)]
    if candidates:
        selected.append(random.choice(candidates))

# Step 2: Add more random names until we reach about 400 total
remaining_needed = 400 - len(selected)
if remaining_needed > 0:
    others = list(set(first_names) - set(selected))
    selected.extend(random.sample(others, remaining_needed))

# Step 3: Shuffle so the alphabet picks are not all at the start
random.shuffle(selected)

# Print or save the selection
print(f"Total selected: {len(selected)}")
for name in selected:
    print(name)

# Optional: save to file
with open("selected_first_names.txt", "w") as out:
    out.write("\n".join(selected))
```
